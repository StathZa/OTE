```python
class Report:
    """Helper for final report"""
    def __init__(self):
        self.lines = []
        self.warnings = []
        self.error = []
        self.start_ts = datetime.now()

    def add(self, line):
        print(f"Appending line to main text:\n{line}")
        self.lines.append(str(line))

    def warn(self, line):
        print(f"Appending line to warnings:\n{line}")
        self.warnings.append(str(line))

    def err(self, line):
        print(f"Appending line to errors:\n{line}")
        self.error.append(str(line))

    def as_text(self):
        out = []
        out.append(f"RUN SUMMARY (Energy Bills import) — started: {self.start_ts:%Y-%m-%d %H:%M:%S}")
        out.append("-" * 70)

        if self.lines:
            out.append("INFO")
            out.extend(f"  - {l}" for l in self.lines)

        if self.warnings:
            out.append("\nWARNINGS")
            out.extend(f"  - {w}" for w in self.warnings)

        if self.error:
            out.append("\nERRORS")
            out.extend(f"  - {e}" for e in self.error)

        out.append("-" * 70)
        out.append(f"finished: {datetime.now():%Y-%m-%d %H:%M:%S}")
        return "\n".join(out)
```

```python
def memory_usage(dataframe):
    '''Function that monitors memory consumption of pandas Dataframe object'''
    if isinstance(dataframe, pd.DataFrame):
        return '{:03.2f} B'.format(dataframe.memory_usage(deep=True).sum())
    else:
        return dataframe.memory_usage(deep=True)
```

```python
SMTP_SERVER = "relay.ote.gr"
SMTP_PORT = 25
SENDER = "EnergyServer@cosmote.gr"


def send_report_text(report_text: str,
                     subject: str,
                     recipients: list,
                     attachment_paths: list = None):
    msg = MIMEMultipart()
    msg["From"] = SENDER
    msg["To"] = ", ".join(recipients)
    msg["Subject"] = subject

    msg.attach(MIMEText(report_text, "plain", _charset="utf-8"))

    for path in (attachment_paths or []):
        try:
            with open(path, "rb") as f:
                part = MIMEApplication(f.read(), _subtype="txt")
            part.add_header("Content-Disposition", f'attachment; filename="{os.path.basename(path)}"')
            msg.attach(part)
        except Exception as e:
            print(f"WARNING: could not attach {path}: {e}")

    with smtplib.SMTP(SMTP_SERVER, SMTP_PORT, timeout=30) as s:
        s.sendmail(SENDER, recipients, msg.as_string())
```

#### Import rows in batches

```python
# Alternative: Explicitly list calculated columns becuase vertica's python driver may not materialize them properly 
# except another interface like ORDER BY enforces them into materialisation 

qry = """
SELECT
    t.*,
    t.hmeromhnia_teleutaias_katametrhshs_new AS last_date,
    t.hmeromhnia_prohgoumenis_katametrhshs_new AS previous_date
FROM (
    SELECT 
        regexp_replace(arithmos_paroxis::varchar, '\\.0', '') AS arithmos_paroxis_new,
        regexp_replace(a_a_ekdosis_logariasmou::varchar, '\\.00', '') AS a_a_ekdosis_logariasmou_new,
        regexp_replace(kwdikos_eett::varchar, '\\.00', '') AS kwdikos_eett_new,
        regexp_replace(kodikos_pollaplou_logariasmou_xxxxxxx::varchar, '\\.0', '') AS kodikos_pollaplou_logariasmou_xxxxxxx_new,
        TO_DATE(REGEXP_SUBSTR(hmeromhnia_teleutaias_katametrhshs::varchar, '^\\d{1,2}/\\d{1,2}/\\d{4}'), 'DD/MM/YYYY') AS hmeromhnia_teleutaias_katametrhshs_new,
        TO_DATE(REGEXP_SUBSTR(hmeromhnia_prohgoumenis_katametrhshs::varchar, '^\\d{1,2}/\\d{1,2}/\\d{4}'), 'DD/MM/YYYY') AS hmeromhnia_prohgoumenis_katametrhshs_new,
        TO_DATE(REGEXP_SUBSTR(logistiko_date::varchar, '^\\d{1,2}/\\d{1,2}/\\d{4}'), 'DD/MM/YYYY') AS logistiko_date_new,
        TO_DATE(REGEXP_SUBSTR(actual_date::varchar, '^\\d{1,2}/\\d{1,2}/\\d{4}'), 'DD/DM/YYYY') AS actual_date_new,
        TO_DATE(REGEXP_SUBSTR(power_off_date::varchar, '^\\d{1,2}/\\d{1,2}/\\d{4}'), 'DD/MM/YYYY') AS power_off_date_new,
        TO_DATE(imera_teleutaiou_logariasmou::varchar(2) || '/' || minas_teleutaiou_logariasmou::varchar(2) || '/' || etos_teleutaiou_logariasmou::varchar(4), 'DD/MM/YYYY') AS Date_Bill,
        *
    FROM energy_efficiency.pollaploi
) t
ORDER BY paroxi ASC, Date_Bill ASC;
"""


batch_size = 10000  
dfs = []
dfs_enanti = []
dfs_ektaktoi = []

try:
    cursor.execute(qry)
    columns = [col[0] for col in cursor.description]

    while True:
        rows = cursor.fetchmany(batch_size)
        if not rows:
            break
        df = pd.DataFrame(rows, columns=columns)
        df.head()
        # perform partial segmentation of bill types
        ektaktoi = df[df["bill_type"] == 'Έκτακτος']
        enanti = df[df["bill_type"] == 'Έναντι']
        df = df.loc[(df["bill_type"] == 'Εκκαθαριστικός') | (df["bill_type"] == 'Τελικός')]
        dfs.append(df)
        dfs_enanti.append(enanti)
        dfs_ektaktoi.append(ektaktoi)
        print(f"Fetched {len(rows)} rows... Total so far: {sum(len(d) for d in dfs)}")
        print(f"{len(ektaktoi)} rows from ektaktoi bills")
        print(f"{len(enanti)} rows from ektaktoi bills")

    df = pd.concat(dfs, ignore_index=True)
    enanti = pd.concat(dfs_enanti, ignore_index=True)
    ektaktoi = pd.concat(dfs_ektaktoi, ignore_index=True)
    print(f"DataFrame bills created successfully. Total rows: {len(df)}")
    print(f"DataFrame enanti created successfully. Total rows: {len(enanti)}")
    print(f"DataFrame ektaktoi created successfully. Total rows: {len(ektaktoi)}")
    del dfs, dfs_ektaktoi, dfs_enanti
    
except Exception as e:
    report.err(f"Error fetching data from pollaploi due to: {e}")
```

```python
cols = ['arithmos_paroxis_new', 'a_a_ekdosis_logariasmou_new', 'kwdikos_eett_new', 
        'kodikos_pollaplou_logariasmou_xxxxxxx_new', 'hmeromhnia_teleutaias_katametrhshs_new',  
        'hmeromhnia_prohgoumenis_katametrhshs_new', 'logistiko_date_new',                  
        'actual_date_new', 'power_off_date_new']
df.drop(columns=[c.replace('_new', '') for c in cols], inplace=True)
enanti.drop(columns=[c.replace('_new', '') for c in cols], inplace=True)
ektaktoi.drop(columns=[c.replace('_new', '') for c in cols], inplace=True)

# Replace the SQL calculated column-names with proper names
cols_dict = {k:k.replace('_new', '') for k in cols}

df.rename(columns=cols_dict, inplace=True)
enanti.rename(columns=cols_dict, inplace=True)
ektaktoi.rename(columns=cols_dict, inplace=True)
```

#### Optimise data types parametrically
* Large DataFrames can lead to significant performance degradation, particularly due to high memory consumption. Optimizing data types can substantially reduce the memory footprint, improve computational efficiency, and enhance overall processing speed when handling large datasets.

```python
def optimize_dataframe(df):
    """Optimize DataFrame memory usage after loading"""
    
    print("Optimizing memory usage...")
    
    # Convert object columns to category
    date_cols = ['paroxi', 'hmeromhnia_teleutaias_katametrhshs', 'hmeromhnia_prohgoumenis_katametrhshs',
                 'logistiko_date', 'actual_date', 'power_off_date', 'last_date', 'previous_date', 'Date_Bill']
    object_cols = df.select_dtypes(include=['object']).columns.difference(date_cols)
    print(object_cols)
    for col in object_cols:
        num_unique = df[col].nunique()
        num_total = len(df)
        
        # Convert to category if < 50% unique values
        if num_unique / num_total < 0.5:
            df[col] = df[col].astype('category')
            print(f"{col}: object → category ({num_unique} unique values)")
    
    # Downcast float64 to float32
    float_cols = df.select_dtypes(include=['float64']).columns
    for col in float_cols:
        df[col] = df[col].astype('float32')
    if len(float_cols) > 0:
        print(f"Converted {len(float_cols)} float64 columns → float32")
        
    return df
```

```python
def datetime_conversion(dataframe):
    date_cols = ['hmeromhnia_teleutaias_katametrhshs', 'hmeromhnia_prohgoumenis_katametrhshs',
                 'logistiko_date', 'actual_date', 'power_off_date', 'last_date', 'previous_date', 'Date_Bill']
    
    for col in date_cols:
        if col in dataframe.columns:
            dataframe[col] = pd.to_datetime(dataframe[col], dayfirst=True, errors='coerce')
            
    return dataframe

df = datetime_conversion(df)
enanti = datetime_conversion(enanti)
ektaktoi = datetime_conversion(ektaktoi)
```

```python
df = df.sort_values(by=["paroxi", "Date_Bill"], ascending=[True, True]).reset_index(drop=True)

max_date=ek[["paroxi", "last_date"]]

max_date = max_date.groupby("paroxi").max().reset_index()

max_date = max_date.rename(columns={"last_date": "last_date_ek"})

max_date['paroxi'] = max_date['paroxi'].astype(int).astype(str)
df['paroxi'] = df['paroxi'].astype(int).astype(str)

df = pd.merge(df, max_date,  how='left', left_on=['paroxi'], right_on=['paroxi'])
print("CHECK 14")
print(df.shape)
df.info()

df.loc[(df['paroxi'] != df['paroxi'].shift(-1)) & (df["bill_type"] == "Έναντι"), 'last_enanti'] = "1"
```

```python
df.loc[(df["last_enanti"] == "1")
       & (df['last_date'] != df['last_date_ek']), 'days_from_last_ek'] = (df['last_date']-df['last_date_ek']).dt.days


df.loc[(df["days_from_last_ek"] > 200), 'alert_kathisterisis'] = "1"
```

#### Alert 2 - Υψηλός Εκκαθαριστικός

```python
median_consumption=ek[["paroxi","daily_consumption"]]

median_consumption=median_consumption.groupby("paroxi").median().reset_index()

median_consumption = median_consumption.rename(columns={"daily_consumption": "median_daily_consumption"})
median_consumption['paroxi'] = median_consumption['paroxi'].astype(int).astype(str)

df = pd.merge(df, median_consumption,
              how='left',
              left_on=['paroxi'],
              right_on=['paroxi'])


df["perc_consumption"]= np.abs(df["daily_consumption"] / df["median_daily_consumption"])
df["perc_consumption"]=df["perc_consumption"].astype(float)


df.loc[((df["daily_consumption"] > 0)
        & (df["median_daily_consumption"] > 0)
        & ((df["bill_type"] == "Εκκαθαριστικός")
           | (df["bill_type"] == "Τελικός"))
        & (df["perc_consumption"] > 2)
        & (df["energy_consumption_kwh"] > 0)
        & (df["correct_parousa_metrisi"]-df["correct_proigoumeni_metrisi"]>20)), 'alert_high_ekkatharistikos'] = "1"
```

#### Alert 3 - Υψηλός Έναντι

```python
df.loc[((df["daily_consumption"] > 0)
        & (df["median_daily_consumption"] > 0)
        & (df["bill_type"] == "Έναντι")
        & (df["perc_consumption"] > 2)
        & (df["energy_consumption_kwh"] > 0)), 'alert_high_enanti'] = "1"
```

#### Alert 4 - Round Consumption

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός")
         | (df["bill_type"] == "Τελικός"))
        & (((df["parousa_endeiksi"] > 0)
            & (df["parousa_endeiksi"] % 1000 == 0))
           | ((df["prohgoumeni_endeiksi"] > 0)
              & (df["prohgoumeni_endeiksi"] % 1000 == 0))
           | ((df["katanalwsh_kwh"] > 0)
              & (df["katanalwsh_kwh"] % 1000 == 0)))
        & (df["katanalwsh_kwh"] > 0)), 'alert_round_numbers'] = "1"
```

#### Alert 5 - Μηδενικές Μετρήσεις

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός")
         | (df["bill_type"] == "Τελικός"))
        & ((df["parousa_endeiksi"] == 0)
           | (df["prohgoumeni_endeiksi"] == 0)
           | (df["katanalwsh_kwh"] == 0)
           | ((df["max_metrisi"] - df["min_metrisi"]) == 0))), 'alert_zero_figures'] = "1"
```

#### Alert 6 - Γύρισμα Μετρητή

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός")
         | (df["bill_type"] == "Τελικός"))
        & (df["parousa_endeiksi"] < df["prohgoumeni_endeiksi"])), 'alert_girisma_metriti'] = "1"
```

#### Alert 7 - Αλλαγή Μετρητή

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός")
         | (df["bill_type"] == "Τελικός"))
        & (df['paroxi']==df['paroxi'].shift(1))
        & (df['arithmos_metriti']!=df['arithmos_metriti'].shift(1))
        & df['arithmos_metriti'].notnull()
        & df['arithmos_metriti'].shift(1).notnull()
        & (df['arithmos_metriti']!="00")
        & (df['arithmos_metriti'].shift(1)!="00")), 'alert_allagi_metriti'] = "1"
```

#### Alert 8 - Λάθος Μέτρηση

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός")
         | (df["bill_type"] == "Τελικός"))
        & (df["synt_wxv"]*(df["correct_parousa_metrisi"]-df["correct_proigoumeni_metrisi"])!=df["katanalwsh_kwh"])),
           'alert_check_consumption'] = "1"
```

#### Alert 9 - Consumption before 2018

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός")
         | (df["bill_type"] == "Τελικός")
         | (df["bill_type"] == "Έναντι") )
        & (df["last_date"].dt.year>=2000)
        & (df["last_date"].dt.year<=2017)
        & (df["previous_date"].dt.year>=2000)
        & (df["previous_date"].dt.year<=2017)), 'alert_old_consumption'] = "1"
```

#### Alert 10 - Zero Consumption

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός") | (df["bill_type"] == "Τελικός"))
        & (df["katanalwsh_kwh"]==0)), 'alert_zero_consumption'] = "1"
```

#### Alert 11 - Low Consumption

```python
df.loc[(((df["bill_type"] == "Εκκαθαριστικός") | (df["bill_type"] == "Τελικός"))
        & (df["perc_consumption"]<=0.4) & (df["perc_consumption"]>0)
        & (df["katanalwsh_kwh"]>0)), 'alert_low_consumption'] = "1"
```

#### Alert 12 - Αλλαγή Παρόχου

```python
df.loc[((df["paroxi"]==df["paroxi"].shift(1))
        & (df["paroxi"].shift(2)==df["paroxi"].shift(1))
        & (df["paroxos"]!=df["paroxos"].shift(1))
        & (df["paroxos"].shift(2)==df["paroxos"].shift(1))
        & (df["bill_type"] != "Έκτακτος")
        & (df["bill_type"].shift(1) != "Έκτακτος")
        & (df["bill_type"].shift(2) != "Έκτακτος")), 'alert_change_provider'] = "1"

df.loc[((df["alert_change_provider"]== "1")
        & (df["paroxi"].shift(1)==df["paroxi"].shift(-1))
        & (df["paroxos"].shift(1)==df["paroxos"].shift(-1))), 'alert_change_provider'] = ""
```

```python
type_mapping = {
    'ote_cosmote': str,
    'paroxos': str,
    'paroxi': str,  # Changed to VARCHAR(255)
    'etos': int,
    'logistikos_minas': int,
    'kodikos_pollaplou_logariasmou_xxxxxxx': str,  # Changed to VARCHAR(255)
    'onoma_pollaplou_1': str,
    'onoma_pollaplou_2': str,
    'perifereia': str,  # Changed to VARCHAR(255)
    'grafeio': str,  # Changed to VARCHAR(255)
    'arithmos_paroxis': str,  # Changed to VARCHAR(255)
    'a_a_ekdosis_logariasmou': str,  # Changed to VARCHAR(255)
    'neos_kodikos_timologiou': str,
    'onoma_pelati': str,
    'dieuthinsi': str,
    'poli_dimos': str,
    'imera_teleutaiou_logariasmou': int,
    'minas_teleutaiou_logariasmou': int,
    'etos_teleutaiou_logariasmou': int,
    'arithmos_metriti': str,
    'eidos_timologiou': str,
    'parousa_endeiksi': float,
    'prohgoumeni_endeiksi': float,
    'synt_wxv': float,
    'imera_teleutaias_katametrisis': int,
    'minas_teleutaias_katametrisis': int,
    'minas_teleutaias_katametrisi': int,
    'etos_teleutaias_katametrisis': int,
    'imera_prohgoumenis_katametrisis': int,
    'minas_prohgoumenis_katametrisis': int,
    'etos_prohgoumenis_katametrisis': int,
    'katanalwsh_kwh': float,
    'axia_energeias_ekdothentos_logariasmou': float,
    'fpa_energeias_ekdothentos_logariasmou': float,
    'ert_ekdothentos_logariasmou': float,
    'logariasmos_kat_ektimisi_enanti_meion': float,
    'axia_energeias_logariasmou_enanti_meion': float,
    'fpa_logariasmou_enanti_meion': float,
    'ert_logariasmou_enanti_meion': float,
    'den_xrhsimopoieitai': str,
    'diafores_xrewseis_pistwseis': float,
    'xrewsh_telous_ape': float,
    'fpa_telous_ape': float,
    'eidikos_foros_katanalwshs': float,
    'poso_dik_ektel_telwn_ergasiwn': float,
    'synolo_xamhlou_fpa': float,
    'synolo_ypshlou_fpa': float,
    'axia_endiamesou': float,
    'synolo_energeias': float,
    'synolo_fpa_reumatos': float,
    'synolo_fpa_yphresiwn': float,
    'synoliko_fpa': float,
    'synolo_ert': float,
    'dhm_telh_dhm_foros': float,
    'telos_akinhths_periousias': float,
    'dosh_eeta': float,
    'plhrwteo_poso': float,
    'synolo_trexontos_mhnos': float,
    'typos_logariasmou': str,
    'hmeromhnia_teleutaias_katametrhshs': 'datetime64[ns]',
    'hmeromhnia_prohgoumenis_katametrhshs': 'datetime64[ns]',
    'energy_cost': float,
    'number_of_days': float,
    'mv_lv': str,
    'energy_consumption_kwh': float,
    'category': str,
    'onomasia': str,
    'site_code_cosmote': str,
    'logistiko_date': 'datetime64[ns]',
    'actual_date': 'datetime64[ns]',
    'kwdikos_eett': str,
    'power_off_date': 'datetime64[ns]',
    'spc_rss': str,
    'bill_type': str,
    'category_new': str,
    'category_group': str,
    'spc': str,
    'nomos': str,
    'perifereia_2': str,
    'diamerisma': str,
    'date_bill': 'datetime64[ns]',
    'last_date': 'datetime64[ns]',
    'previous_date': 'datetime64[ns]',
    'number_of_days2': int,
    'min_metrisi': float,
    'max_metrisi': float,
    'correct_proigoumeni_metrisi': float,
    'correct_parousa_metrisi': float,
    'delta_days': int,
    'daily_consumption': float,
    'new_bill_code': str,
    'siteid': str,
    'latitude': float,
    'longitude': float,
    'last_date_ek': 'datetime64[ns]',
    'last_enanti': str,
    'days_from_last_ek': float,
    'alert_kathisterisis': str,
    'median_daily_consumption': float,
    'perc_consumption': float,
    'alert_high_ekkatharistikos': str,
    'alert_high_enanti': str,
    'alert_round_numbers': str,
    'alert_zero_figures': str,
    'alert_girisma_metriti': str,
    'alert_allagi_metriti': str,
    'alert_check_consumption': str,
    'alert_old_consumption': str,
    'alert_zero_consumption': str,
    'alert_low_consumption': str,
    'alert_change_provider': str,
    'alarm_priority': float,
    'par_dt': str
}

def convert_value(value, target_type):
    """Convert a single value to the expected Vertica-compatible type."""
    if pd.isna(value):
        return None

    # --- handle categories, numpy types, etc. ---
    if isinstance(value, (np.generic, np.bool_)):
        value = np.asscalar(value) if hasattr(np, "asscalar") else value.item()

    # --- dispatch based on target type ---
    if target_type in [str]:
        return str(value)
    elif target_type in [int]:
        try:
            return int(float(value))
        except Exception:
            return 0
    elif target_type in [float]:
        try:
            return float(value)
        except Exception:
            return None
    elif target_type == "datetime64[ns]":
        if isinstance(value, (datetime, pd.Timestamp)):
            return value
        try:
            return pd.to_datetime(value, errors="coerce")
        except Exception:
            return None
    else:
        # fallback: return as string
        return str(value)


def insert_data(cursor, connection, type_mapping,
                df: pd.DataFrame = df, 
                schema: str = schema, 
                target_table: str = target_table,  
                batch_size=50000) -> None:
    cols_list = [f'"{col}"' for col in df.columns]
    cols_str = ", ".join(cols_list)
    placeholders = ", ".join(["%s"] * len(df.columns))
    insert_sql = f"INSERT INTO {schema}.{target_table} ({cols_str}) VALUES ({placeholders})"

    total_inserted = 0
    print(f"Starting insertion of {len(df):,} rows into {schema}.{target_table}...")

    for i in range(0, len(df), batch_size):
        batch = df.iloc[i:i + batch_size]
        batch_data = []

        for _, row in batch.iterrows():
            converted_row = []
            for col, val in zip(df.columns, row.values):
                # get correct dtype from mapping
                target_type = type_mapping.get(col.lower(), str)
                conv_val = convert_value(val, target_type)
                converted_row.append(conv_val)
            batch_data.append(tuple(converted_row))

        cursor.executemany(insert_sql, batch_data)
        connection.commit()
        total_inserted += len(batch_data)

        print(f"Batch {i // batch_size + 1}: inserted {len(batch_data):,} rows (total: {total_inserted:,})")

    print(f"\nINSERT complete! Total rows inserted: {total_inserted:,}")

insert_data(
    cursor,
    connection,
    type_mapping=type_mapping,
    batch_size=20000
)
```
